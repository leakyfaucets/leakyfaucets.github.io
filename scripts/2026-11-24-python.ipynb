### Python

当变量被赋值时，Python会根据对象的类型决定是传递引用还是创建拷贝。对于可变对象（如列表、字典），赋值操作会传递引用，因此修改变量会影响原始对象。对于不可变对象（如字符串、元组、数字），赋值操作会创建新的对象。

```python
# 可变对象的引用
dict_obj = {'a': [1, 2, 3], 'b': 2}
x = dict_obj['a']
x[0] = 99
print(dict_obj) # 输出: {'a': [99, 2, 3], 'b': 2}

# 不可变对象的赋值
y = dict_obj['b']
y += 1
print(dict_obj) # 输出: {'a': [99, 2, 3], 'b': 2}
```

浅拷贝会生成一个新的对象，但对象内部的可变子对象（如列表中的列表、字典中的列表等）不会被复制，而是共享原对象的引用。这意味着修改嵌套的可变对象会影响原对象，而修改顶层对象不会影响原对象。

python中常用的浅拷贝方法包括列表切片、`list`和`dict`函数、列表和字典的`copy()`方法，以及`copy`模块的`copy`函数。

深拷贝递归复制对象及其所有嵌套对象，生成完全独立的新对象，修改新对象不会影响原对象。深拷贝通常使用`copy`模块的`deepcopy`函数实现。

```python
import copy

lst = [[1, 2, 3], (4, 5)]
lst1 = lst[:]
lst2 = lst.copy()
lst3 = copy.copy(lst)
lst4 = copy.deepcopy(lst)
def test(x, y):
    x[0][0] = 0
    x[1] = 0
    print(y)
test(lst4, lst)
test(lst1, lst)
test(lst2, lst)
test(lst3, lst)
```

### numpy

numpy中axis从最外层开始排列。如np.array([[1,4,8],[2,3,5],[2,5,1],[1,10,7]])的第零维大小为4，第一维的大小为3。

```python
import numpy as np

item = np.array([[1,4,8],[2,3,5],[2,5,1],[1,10,7]])
print(np.size(item, 0))
print(np.size(item, 1))
print(item.shape) # (4,3)
print(item.sum(axis=0)) # 沿第零维求和，结果为(3,)的数组
print(item.sum(axis=1)) # 沿第一维求和，结果为(4,)的数组
```

numpy数组的维度用元组保存，因此一维数组需要表示为(i,)，(i)不是元组。

当两个形状不同的数组进行对应元素操作（如加减乘除，不包括点积）时会发生广播。只有当两个数组各维度大小从后往前比对均一致，或不一致时有一个数组的该维度大小为1时，才能进行广播。

```python
# 最后一维不一致且均不为1，无法广播
A = np.zeros((2,5,3,4))
B = np.zeros((3,3))
try: print((A+B).shape)
except Exception as e: print(e)

# 最后一维不一致但有一个为1，可以广播
A = np.zeros((2,5,3,4))
B = np.zeros((3,1))
print((A+B).shape)

# 第二维不一致且均不为1，无法广播
A = np.zeros((2,5,3,4))
B = np.zeros((2,4,1,4))
try: print((A+B).shape)
except Exception as e: print(e)
```
