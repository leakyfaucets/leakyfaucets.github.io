## 基础语法

JS的每个语句由分号结尾，语句块由大括号括起来。语句块内的语句通常进行4个空格的缩进，如：

```python
%%javascript
if (2 > 1) {
    x = 1;
    y = 2;
    z = 3;
}
```

单行注释以`//`开头，多行注释由`/*`和`*/`包裹，如：

```python
%%javascript
/* 从这里开始是块注释
仍然是注释
仍然是注释
注释结束 */
```

Javascript严格区分大小写。

### 变量

变量名通常使用拉丁字母、数字和下划线的组合。注意变量名不能以数字和下划线开头，也不能使用保留关键字。

使用`let`语句声明本地变量并初始化，如`let myname=Chris`。若不使用`let`语句，将会声明全局变量。

```python
%%javascript
// 打印变量x
var x = 100;
console.log(x);
```

### 常量

使用`const`关键字来声明本地常量。

### 解构赋值

可以同时对多个变量进行赋值。对数组元素进行解构赋值时，多个变量要用中括号括起来。解构赋值还可以忽略某些元素。

```python
%%javascript
let [x, y, z] = ['hello', 'JavaScript', 'ES6'];
element.text(`x = ${x}, y = ${y}, z = ${z}`);
let [, , a] = ['hello', 'JavaScript', 'ES6']; // 忽略前两个元素，只对z赋值第三个元素
a; // 'ES6'
```

对象也可进行解构赋值，变量要用大括号括起来。

```python
%%javascript

let person = {
    name: '小明',
    age: 20,
    gender: 'male',
    passport: 'G-12345678',
    school: 'No.4 middle school'
};
let {name, age, passport} = person;

// name, age, passport分别被赋值为对应属性:
element.text(`name = ${name}, age = ${age}, passport = ${passport}`);
```

如果要使用的变量名和属性名不一致，可以用下面的语法获取：

```python
%%javascript

let person = {
    name: '小明',
    age: 20,
    gender: 'male',
    passport: 'G-12345678',
    school: 'No.4 middle school'
};

// 把passport属性赋值给变量id:
let {name, passport:id} = person;
name; // '小明'
id; // 'G-12345678'
// 注意: passport不是变量，而是为了让变量id获得passport属性:
passport; // Uncaught ReferenceError: passport is not defined
```

解构赋值还可以使用默认值，这样就避免了不存在的属性返回`undefined`的问题：

```python
%%javascript
let person = {
    name: '小明',
    age: 20,
    gender: 'male',
    passport: 'G-12345678'
};

// 如果person对象没有single属性，默认赋值为true:
let {name, single=true} = person;
name; // '小明'
single; // true
```

如果变量已经被声明了，解构赋值时需要用小括号括起来赋值语句：

```python
%%javascript

let x, y;
{x, y} = { name: '小明', x: 100, y: 200};
({x, y} = { name: '小明', x: 100, y: 200}); // 注意必须把语句放在小括号里，否则会被当成代码块
```

### 数据类型

`null`表示空值，`undefined`表示未定义。

#### 数字

JS不区分整数和浮点数，以下数值均合法：

- 123; // 十进制整数
- 0xff00; // 十六进制整数
- 0.456; // 浮点数
- 1.2345e3; // 科学计数法表示
- -99; // 负数
- NaN; // 当无法计算结果时用NaN表示
- Infinity; // 当数值超过了Number所能表示的最大值（$±2^{53}$）时，就表示为Infinity

数字运算符与Python一致。自增运算符为`++`，自减运算符为`--`。`+=`、`-=`、`*=`和`/=`运算符效果与Python一致。

#### 字符串

字符串用一对单引号或双引号括起来。JS字符串同样可以使用转义字符。多行字符串使用两个`` ` ``包裹起来，也可在换行处使用`\n`。字符串和变量可以通过`+`连接，也可使用模板字符串，用两个`` ` ``包裹起来，其中变量使用`${variant}`格式。

```python
%%javascript
// 方式一：转义字符 \n
const a = "这是一个\n多行\n字符串";

// 方式二：直接换行
const b = `这是一个
多行
字符串`;
element.html(`<pre>${b}</pre><pre>${a}</pre>`);
```

JS字符串同样支持索引，获取字符串长度使用length属性。注意：字符串不可变，如果对字符串的某个索引赋值，不会有任何错误，也没有任何效果。常用方法见下。

```python
%%javascript
let s = 'Hello, world!';
s.length; // 13
s[0]; // 'H'
s[13]; // undefined 超出范围的索引不会报错，但会返回undefined
s.toUpperCase(); // 返回'HELLO, WORLD!'
s.toLowerCase(); // 返回'hello, world!'
s.indexOf('world'); // 返回7
s.indexOf('World'); // 没有找到指定的子串，返回-1
s.substring(0, 5); // 从索引0开始到5（不包括5），返回'hello'
s.substring(7); // 从索引7开始到结束，返回'world'
```

#### 数组

JS的数组与Python的列表形式相同，也可以通过索引来访问。length属性返回数组长度。大多数其他编程语言不允许直接改变数组的大小，且越界访问索引会报错。然而，JS却不会有任何错误。在编写代码时，不建议直接修改Array的大小，访问索引时要确保索引不会越界。

```python
%%javascript
var arr = [1, 2, 3.14, 'Hello', null, true];
arr[0]; // 返回索引为0的元素，即1
arr[6]; // 索引超出了范围，返回undefined
arr.length; // 返回数组的长度，即6
arr.length = 4; // 将数组长度设置为4，超出部分会被删除
arr; // 返回[1, 2, 3.14, 'Hello']
arr.length = 6; // 将数组长度设置回6，新增的元素会是undefined
arr; // 返回[1, 2, 3.14, 'Hello', undefined, undefined]
arr[0] = 100; // 修改索引0的元素为100
arr; // 返回[100, 2, 3.14, 'Hello', undefined, undefined]
arr[7] = 200; // 在索引7处添加一个新元素200
arr; // 返回[100, 2, 3.14, 'Hello', undefined, undefined, undefined, 200]
```

常用方法有：
- indexOf()：搜索一个指定元素在数组中的位置。
- slice()：截取数组的部分元素并返回一个新数组。
- unshift()：向数组的开头添加若干元素；push()：向数组的末尾添加若干元素。
- shift()：删除数组的第一个元素；pop()：删除数组的最后一个元素。
- sort()：对元素进行排序，可指定顺序。
- reverse()：反转元素。
- splice()：从指定的索引开始删除若干元素，然后再从该位置添加若干元素。
- concat()：接收任意个元素和数组，并且把其中的数组展开，然后全部添加到新的数组里。
- join()：把每个元素用指定的字符串连接起来，并返回连接后的字符串。

```python
%%javascript
let arr = [10, 20, '30', 'xyz'];
arr.indexOf(10); // 元素10的索引为0
arr.indexOf(30); // 元素30没有找到，返回-1
arr = ['A', 'B', 'C', 'D', 'E', 'F', 'G'];
arr.slice(0, 3); // 从索引0开始，到索引3结束，但不包括索引3: ['A', 'B', 'C']
arr.slice(3); // 从索引3开始到结束: ['D', 'E', 'F', 'G']
let aCopy = arr.slice(); // 如果不给slice()传递任何参数，它就会从头到尾截取所有元素。利用这一点，我们可以很容易地复制一个数组
aCopy; // ['A', 'B', 'C', 'D', 'E', 'F', 'G']
aCopy === arr; // false aCopy和arr是两个不同的数组对象

arr = [1, 2];
arr.push('A', 'B'); // 返回Array新的长度: 4
arr; // [1, 2, 'A', 'B']
arr.pop(); // pop()返回'B'
arr; // [1, 2, 'A']
arr.pop(); arr.pop(); arr.pop(); // 连续pop 3次
arr; // []
arr.pop(); // 空数组继续pop不会报错，而是返回undefined
arr; // []

let arr = ['Microsoft', 'Apple', 'Yahoo', 'AOL', 'Excite', 'Oracle'];
// 从索引2开始删除3个元素,然后再添加两个元素:
arr.splice(2, 3, 'Google', 'Facebook'); // 返回删除的元素 ['Yahoo', 'AOL', 'Excite']
arr; // ['Microsoft', 'Apple', 'Google', 'Facebook', 'Oracle']
// 只添加,不删除:
arr.splice(2, 0, 'Google', 'Facebook'); // 返回[],因为没有删除任何元素
arr; // ['Microsoft', 'Apple', 'Google', 'Facebook', 'Oracle']

let arr = ['A', 'B', 'C'];
arr.concat(1, 2, [3, 4]); // ['A', 'B', 'C', 1, 2, 3, 4]
```

#### 对象

JS的对象与Python的字典形式类似，也为无序键值对，不同的是，JS的键为字符串，不需要用引号括起来。

```python
%%javascript
var person = {
    name: 'Bob',
    age: 20,
    tags: ['js', 'web', 'mobile'],
    city: 'Beijing',
    hasCar: true,
    zipcode: null
};
element.text(`person.name=${person.name}, person.age=${person.age}, person.tags=${person.tags}, person.city=${person.city}, person.hasCar=${person.hasCar}, person.zipcode=${person.zipcode}`);
```

要获取一个对象的属性，需要用`变量名.属性名（键名）`的方式。访问不存在的属性不报错，而是返回`undefined`。

#### 集合

JS的集合满足互异性，但不满足无序性（保持插入顺序）。创建集合既可以提供一个数组作为输入，也可以直接创建空集合。可以通过`add()`方法和`delete()`方法增加或删除元素。

```python
%%javascript
let s1 = new Set(); // 空Set
let s = new Set([1, 2, 3, 3, '3']); // Set构造函数会自动去重
s; // Set {1, 2, 3, "3"}
s.delete("3"); // 删除元素'3'
s; // Set {1, 2, 3}
s.add(4); // 添加元素4
s; // Set {1, 2, 3, 4}
```

#### 映射

映射相当于增强型的对象，其键可以是任意数据类型，同时也保持了插入序。创建映射需要一个二维数组，或者直接创建一个空映射。映射具有以下方法：

```python
%%javascript
let m1 = new Map([['Michael', 95], ['Bob', 75], ['Tracy', 85]]);
m1.get('Michael'); // 95
let m = new Map(); // 空Map
m.set('Adam', 67); // 添加新的key-value
m.set('Bob', 59);
m.has('Adam'); // 是否存在key 'Adam': true
m.get('Adam'); // 67
m.delete('Adam'); // 删除key 'Adam'
m.get('Adam'); // undefined
m.set('Adam', 88);
m.get('Adam'); // 88, 同一个键被重新赋值
```

### 条件语句

JS的单条件语句格式为`if (条件) {条件为真时执行的语句} else {条件为假时执行的语句}`，其中`else`语句是可选的。嵌套条件可使用`else if (条件) {...}`语句。JS把`null`、`undefined`、`0`、`NaN`和`''`视为`false`，其他值视为`true`。

```python
%%javascript
let age = 3;
if (age >= 18) {
    element.text('adult');
} else if (age >= 6) {
    element.text('teenager');
} else {
    element.text('kid');
}
```

### 循环语句

#### for循环

`for (初始化器（计数器、迭代器）; 条件; 最终语句) {语句}`。小括号内的三个属性都是可选的，如果没有退出循环判断条件，需要使用`break`语句退出循环。`continue`语句可跳过当前循环。

```python
%%javascript
let x = 0;
let i;
for (i=1; i<=10000; i++) {
    x = x + i;
}
element.text(x); // 50005000
```

`for...in...`语句可以遍历一个对象的所有属性。要过滤掉对象继承的属性，用`hasOwnProperty()`方法来实现。注意迭代器变量均要用显式声明。

```python
%%javascript
let o = {
    name: 'Jack',
    age: 20,
    city: 'Beijing'
};
for (let key in o) {
    if (o.hasOwnProperty(key)) {
        console.log(key); // 'name', 'age', 'city'
    }
}
```

由于数组也是对象，它的每个元素的索引被视为对象的属性，因此，`for...in...`语句可以直接循环数组的索引，得到的是**字符串**索引。对于数组，更推荐使用`for...of...`语句，会直接遍历可迭代对象的值。

```python
let arr = ['Bart', 'Lisa', 'Adam'];
for (let i of arr) {
    console.log(`Hello, ${i}!`)
}; // Hello, Bart! Hello, Lisa! Hello, Adam!
```

数组、映射和集合都属于可迭代对象。它们有`forEach()`方法，可以同时遍历属性、索引、值等。

```python
%%javascript
let m = new Map([[1, 'x'], [2, 'y'], [3, 'z']]);
m.forEach(function (value, key) {
    element.text(key + ': ' + value);
});
```

#### while循环

`while (条件) {语句}`在循环开始时判断条件；`do {语句} while (条件);`在循环结束时判断条件，因此循环体至少会执行1次。请<mark>注意</mark>：仅有`do {语句} while (条件);`需要在结尾加分号，其他循环/条件语句均不需要，因为这些语句属于复合语句，以其代码块的大括号为语法终结符。

```python
%%javascript
let n = 0;
do {
    n = n + 1;
} while (n < 100);
n; // 100
```

### 函数

JS通过`function func(x1, x2) {}`语句定义函数。如果没有`return`语句，函数执行完毕后会返回`undefined`。JS允许传入任意个参数而不影响调用。`arguments`关键字指向所有传入的参数，常用其`length`属性判断传入参数的个数。同样可以使用默认参数。

```python
%%javascript

function foo(a, b, c = null) {
    element.text(`${a}, ${b}, ${c}`)
}
foo(1, 2); // 1, 2, null
```

`...rest`参数可用于获取其他参数，须放在所有其他参数之后。多余的参数将以数组形式传给变量`rest`。

```python
%%javascript

function foo(a, b, ...rest) {
    console.log('a = ' + a);
    console.log('b = ' + b);
    console.log(rest);
}

foo(1, 2, 3, 4, 5);
// 结果:
// a = 1
// b = 2
// Array [ 3, 4, 5 ]

foo(1);
// 结果:
// a = 1
// b = undefined
// Array []
```
