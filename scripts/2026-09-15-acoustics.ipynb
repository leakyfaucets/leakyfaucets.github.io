电磁波为横波，机械波包括纵波和横波。声波为在弹性介质中传播的机械纵波。横波：质点的运动方向与波的传播方向垂直。纵波：方向一致，如弹簧产生的波。

正弦波：$D=A\sin{\frac{2\pi t}{T}}=A\sin{(2\pi f t)}=A\sin{(\omega t)}$，角频率$\omega=2\pi f=\frac{2\pi}{T}$

相位（phase）：$D=A\sin{\frac{2\pi t}{T}+\phi}=A\sin{(2\pi f t+\phi)}=A\sin{(\omega t+\phi)}$，$\phi$：初相位
